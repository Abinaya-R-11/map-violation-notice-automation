```python
import os
import pandas as pd
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
```

```python
violation = pd.read_csv("top_violations.csv")
violation
```

```python
violation["price_diff"] = round(violation["LPP"]-violation["advertised_price"],2)
```

```python

for index,x in violation.iterrows():
    filename = f"MAP_notice_{index}_{x['seller_name']}_{x['sku']}.pdf"

    email_letter = f""" OFFICIAL NOTICE: MINIMUM ADVERTISED PRICE (MAP) COMPLIANCE VIOLATION

Date: {x["violation_date"]}
Seller Name: {x["seller_name"]}
Sales Platform: {x["marketplace"]}
Region: {x["region"]}
Product Line: {x["PL"]}
SKU: {x["sku"]}

SUBJECT: Formal Notice of Pricing Policy Violation - SKU: {x["sku"]}

Dear {x["seller_name"]} Team,

During an automated price monitoring audit, an active listing under your account was identified as non-compliant with our
Minimum Advertised Price(MAP) policy.

VIOLATION SPECIFICATIONS:

- Product SKU: {x["sku"]}
- Product Line: {x["PL"]}
- LPP: ${x["LPP"]}
- Advertised Listing Price: ${x["advertised_price"]}
- Price Variance: ${x["price_diff"]}
- Violation Classification: {x["violation_type"]}

REQUIRED CORRECTIVE ACTION:

1. Adjust your public listing price on {x["marketplace"]}({x["region"]}) to at or above ${x["LPP"]}.
2. Changes must propagate across the marketplace within 24 hours of this notice.
3. Reply directly to this notification once the listing update is active.

Failure to remediate this price within 24 hours may result in temporary suspension of product authorization and partner 
privileges.

Sincerely,
Brand Protection & Pricing Compliance Operations """

    c = canvas.Canvas(filename,pagesize=letter)
    c.setFont("Helvetica",10)

    text_obj = c.beginText(40,750)
    for line in email_letter.split("\n"):
        text_obj.textLine(line)
    c.drawText(text_obj)

    c.save()
    os.startfile(filename)
    
```
