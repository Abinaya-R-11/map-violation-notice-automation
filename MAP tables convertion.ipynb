## populate price list table

### calculate LPP

```python
df_price = pd.read_csv("price_list_table.csv")
df_price
```

```python
valid_rows  = (df_price["PL"].apply(lambda x:subcat_map.get(str(x).split("_")[-1].strip(),"unknown")) != "unknown")
```

```python
df_price = df_price[valid_rows]
```

```python
df_price
```

```python
usd_rate = 83.0

df_price = df_price.copy()

df_price["MAP_usd"] = (df_price["MAP"]/usd_rate).round(2)

```

```python
df_price
```

```python
discounts = { ("HP","Laptop"): 0.034,
              ("DELL","Laptop"):0.078,
              ("HP","Monitor"): 0.012,
              ("DELL","Monitor"):0.056,
              ("HP","Toner"): 0.091,
              ("DELL","Toner"):0.023,
              ("HP","Laserjet"): 0.065,
              ("DELL","Laserjet"):0.035,
              ("HP","Inkjet"): 0.012,
              ("DELL","Inkjet"):0.056,
              ("HP","Desktop"): 0.091,
              ("DELL","Desktop"):0.023,
              ("HP","Ink"): 0.065,
              ("DELL","Ink"):0.035}
```

```python
def calculate_lpp(x):
    pl = x["PL"]
    brand = "HP" if pl.startswith("HP") else "DELL"
    ending = pl.split("_")[-1].strip()
    subcat = subcat_map[ending]
    discount_rate = discounts[(brand,subcat)]
    return round(x["MAP_usd"] * (1-discount_rate),2)
```

```python
df_price["LPP"] = df_price.apply(calculate_lpp,axis=1)
```

```python
df_price_table = df_price[["PL","sku","MAP","MAP_usd","LPP"]].drop_duplicates()
```

```python
df_price_table
```

```python
df_price_table.to_csv("price_list_table_new_2.csv",index = False)
```

## populate promotion table

```python
df_promotional_sku = pd.read_excel("promotional_skus.xlsx")
df_promotional_sku.to_csv("promotional_skus.csv",index = False)
```

```python
df_promotional_sku
```

```python
promo_list =(df_promotional_sku["sku"].astype(str).str.strip().tolist())
```

```python
df_promo = df_price[df_price["sku"].isin(promo_list)].copy()
df_promo = pd.merge(df_promo,df_cat[["sku","category","sub_category"]],on="sku",how="left")
```

```python
df_promo
```

```python
def calculate_q4_promo(x):
    cat = x["category"]
    subcat = x["sub_category"]
    price = x["MAP_usd"]


    if cat =="PC":
        return round(price*(1-0.15),2)
    elif cat == "Supply":
        if "Ink" in subcat:
            return round(price*(1-0.20),2)
    elif cat == "Print Hardware":
        if "Laserjet" in subcat:
            discount = 10.0
            return round(price-discount,2)
    return round(price,2)
```

```python
df_promo["season"] = "Q4"
df_promo["promotion"] = df_promo.apply(calculate_q4_promo,axis = 1)
```

```python
df_promo
```

```python
df_promo_table = df_promo[["PL","sku","season","promotion"]]
df_promo_table
```

```python
df_promo_table.to_csv("promotion_table_new.csv",index = False)
```

## seller table

```python
import glob
import os
```

```python
files = glob.glob(os.path.join("Seller Data","*.csv"))
print(f"Total files found:,{len(files)}")
print("Sample files",files[:3])
```

```python
collected_dfs = []
for file in files:
    df = pd.read_csv(file)
    date_label=os.path.basename(file).replace(".csv","")
    df["last_time_collected"] = date_label
    collected_dfs.append(df)

combined_seller_df = pd.concat(collected_dfs,ignore_index=True)
len(combined_seller_df)
```
